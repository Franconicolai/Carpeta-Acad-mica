# Función que permite armar un diccionario vacío con las probabilidades de transición y la lista de estados (NO MODIFICAR)

```python
def armar_estados(lista_estaciones, lista_capacidad):
    estados = []
    for i in lista_capacidad:
        for j in lista_estaciones:   
            estados.append((i, j, 0))
            estados.append((i, j, 1))
    
    # Armar diccionario con las probabilidades de transición, seteadas en 0:
    probabilidades_transicion = {}
    for s1 in estados:
        for s2 in estados:
            probabilidades_transicion[(s1, s2)] = 0
    
    return estados, probabilidades_transicion




    
```

# PARAMETROS DE SUBIDA Y BAJADA DE PASAJEROS (NO MODIFICAR)

```python
def armar_parametros_variables_aleatorias(lista_estaciones, capacidad_maxima):
    np.random.seed(2123)
    parametros_subida = {}
    parametros_bajada = {}
    for i in lista_estaciones:
        parametros_bajada[i] = round(np.random.uniform(0.1, 0.9),1)
        if i != lista_estaciones[-1]:
            parametros_subida[i, i+1] = int(round(np.random.uniform(2, capacidad_maxima),0))
        if i != lista_estaciones[0]:
            parametros_subida[i, i-1] =  int(round(np.random.uniform(2, capacidad_maxima),0))
    
    return parametros_subida, parametros_bajada
```

# Función que permite obtener probabilidad de subida al viajar de una estación i a una j. (NO MODIFICAR)

```python
def obtener_probabilidad_exacta_subida(parametro_subida, parametro_bajada, cantidad_actual,  cantidad_nueva, estacion_actual, estacion_siguiente):
    min_uniforme = 0
    max_uniforme = parametro_subida[(estacion_actual, estacion_siguiente)]
    prob_binomial = parametro_bajada[estacion_siguiente]
    prob_total = 0
    for i in range(cantidad_actual+1):
        if i + cantidad_nueva - cantidad_actual > max_uniforme:
            break
        prob_uniforme_discreta_i = 1/(max_uniforme - min_uniforme +1)
        prob_binomial_i = math.comb(cantidad_actual, i) * (prob_binomial**i) * ((1-prob_binomial)**(cantidad_actual-i))
        prob_total += prob_uniforme_discreta_i * prob_binomial_i
    return prob_total
```

# Función que permite obtener la probabilidad de que el vagón se llene. (NO MODIFICAR)

```python
def probabilidad_hasta_m(parametro_subida, parametro_bajada, cantidad_actual, capacidad_maxima, estacion_actual, estacion_siguiente):
    min_uniforme = 0
    max_uniforme = parametro_subida[(estacion_actual, estacion_siguiente)]
    prob_binomial = parametro_bajada[estacion_siguiente]
    prob_total = 0
    for i in range(cantidad_actual+1):
        prob_uniforme_discreta_i = (1- (capacidad_maxima - cantidad_actual +i  - min_uniforme)/(max_uniforme - min_uniforme +1))
        prob_binomial_i = math.comb(cantidad_actual, i) * (prob_binomial**i) * ((1-prob_binomial)**(cantidad_actual-i))
        if prob_binomial_i < 0 or prob_uniforme_discreta_i < 0:
            break
        prob_total += prob_uniforme_discreta_i * prob_binomial_i


    return prob_total
```

# Función que permite obtener la probabilidad de que el vagón baje la cantidad de personas al viajar de una estación a otra. (NO MODIFICAR)

```python
def obtener_probabilidad_exacta_bajada(parametro_subida, parametro_bajada, cantidad_actual,  cantidad_nueva, estacion_actual, estacion_siguiente):
    min_uniforme = 0
    max_uniforme = parametro_subida[(estacion_actual, estacion_siguiente)]
    prob_binomial = parametro_bajada[estacion_siguiente]
    prob_total = 0
    for i in range(max_uniforme+1):
        
        if i > cantidad_nueva:
            break
        prob_uniforme_discreta_i = 1/(max_uniforme - min_uniforme +1)
        prob_binomial_i = math.comb(cantidad_actual, cantidad_actual - cantidad_nueva + i) * (prob_binomial**(cantidad_actual - cantidad_nueva + i)) * ((1-prob_binomial)**(cantidad_nueva-i))
        prob_total += prob_uniforme_discreta_i * prob_binomial_i
    
    return prob_total
```

# Permite rellenar la matriz de probabilidades de transición. (NO MODIFICAR)

```python
def armar_matriz_de_probabilidad_transicion(estados, probabilidades_transicion, parametros_subida, parametros_bajada, capacidad_maxima):
    lista_estados_out = []
    
    for s1 in estados:
        c, e, d = s1

        if d== 0 and e == cantidad_estaciones:
            lista_estados_out.append(s1)
            continue
            
        if d == 1 and e == 1:
            lista_estados_out.append(s1)
            continue

        for c_prima in capacidades:
            
            if d == 1:
                if e == 2:
                    proximo_e = 1
                    proximo_d = 0
                elif e > 2:
                    proximo_e = e - 1
                    proximo_d = 1
            else:
                if e == cantidad_estaciones-1:
                    proximo_e = cantidad_estaciones
                    proximo_d = 1
                elif e < cantidad_estaciones-1:
                    proximo_e = e + 1
                    proximo_d = 0
                     

                

            s2 = (c_prima, proximo_e, proximo_d)
            if c_prima < c:
                prob = obtener_probabilidad_exacta_bajada(parametros_subida, parametros_bajada, c, c_prima, e, proximo_e)
                
                probabilidades_transicion[(s1, s2)] = prob
            else:
                if c_prima - c > parametros_subida[(e, proximo_e)]:
                    break
                if c_prima == capacidad_maxima:
                    prob = probabilidad_hasta_m(parametros_subida, parametros_bajada, c, capacidad_maxima, e, proximo_e)
                    probabilidades_transicion[(s1, s2)] = prob

                else:
                    prob = obtener_probabilidad_exacta_subida(parametros_subida, parametros_bajada, c, c_prima, e, proximo_e)
                    probabilidades_transicion[(s1, s2)] = prob
    
    # Quitar los estados out de la lista de estados (estadps inancanzables) y de las probabilidades de transición:
    for s in lista_estados_out:
        estados.remove(s)
        for s2 in estados:
            probabilidades_transicion.pop((s, s2))
            probabilidades_transicion.pop((s2, s))

    return probabilidades_transicion, estados

                
                    
```

# Función que utiliza DFS y memoización para contar caminos. A partir de este código, deben inspirarse para implementar lo que se solicita en el apartado b (NO MODIFICAR)

```python
# Función para contar e imprimir caminos válidos en una grilla con obstáculos usando DFS + memoización
def contar_e_imprimir_caminos(m, n, celdas_bloqueadas):
    bloqueadas = set(celdas_bloqueadas)
    memo = {}

    def dfs(i, j, path):
        # Si estamos fuera de la grilla o en una celda bloqueada
        if i >= m or j >= n or (i, j) in bloqueadas:
            return 0

        # Si llegamos al destino, imprimir el camino
        if (i, j) == (m-1, n-1):
            path.append((i, j))
            print("→ Camino:", path)
            path.pop()
            return 1

        if (i, j) in memo:
            # No usamos memo aquí porque los caminos pueden variar incluso con el mismo número total
            # Este memo se puede usar solo para conteo, no para caminos explícitos
            pass

        path.append((i, j))
        total = 0
        total += dfs(i+1, j, path)  # Mover abajo
        total += dfs(i, j+1, path)  # Mover derecha
        path.pop()

        memo[(i, j)] = total
        return total

    total_caminos = dfs(0, 0, [])
    print(f"\n🔢 Total de caminos posibles: {total_caminos}")

# Ejemplo de uso
m = 5
n = 5
celdas_bloqueadas = [(1, 1), (3,2)]

contar_e_imprimir_caminos(m, n, celdas_bloqueadas)
```

## Inciso b)

```python
def calcular_probabilidad(estacion_inicial, estacion_final, personas_iniciales, personas_finales):
    # Determinar la dirección inicial según el movimiento requerido
    # Si vamos de estacion_inicial a estacion_final y estacion_final > estacion_inicial, asumimos que la dirección es 0 (avance);
    # caso contrario, si estacion_final < estacion_inicial, la dirección es 1 (retroceso)
    if estacion_final > estacion_inicial:
        direccion_inicial = 0
    else:
        direccion_inicial = 1
    estado_inicial = (personas_iniciales, estacion_inicial, direccion_inicial)
    cantidad_viajes = abs(estacion_inicial - estacion_final)
    
    memoizacion = {}
    def calcular_probabilidad_recursiva(estado_actual, viajes_restantes):
        if viajes_restantes == 0:
            # Se evalúa si se llegó a la estación destino estacion_final y con personas_finales personas.
            if estado_actual[1] == estacion_final and estado_actual[0] == personas_finales:
                return 1
            return 0
        if (estado_actual, viajes_restantes) in memoizacion:
            return memoizacion[(estado_actual, viajes_restantes)]
        probabilidad_acumulada = 0
        for siguiente_estado in estados:
            probabilidad_transicion = probabilidades_transicion.get((estado_actual, siguiente_estado), 0)
            if probabilidad_transicion > 0:
                probabilidad_acumulada += probabilidad_transicion * calcular_probabilidad_recursiva(siguiente_estado, viajes_restantes-1)
        memoizacion[(estado_actual, viajes_restantes)] = probabilidad_acumulada
        return probabilidad_acumulada
    
    return calcular_probabilidad_recursiva(estado_inicial, cantidad_viajes)

# Definir los pares de estaciones según lo solicitado
cantidad_estaciones_totales = cantidad_estaciones  
pares_estaciones = [
    (1, cantidad_estaciones_totales),
    (cantidad_estaciones_totales, 1),
    (1, cantidad_estaciones_totales//2),  # ⌊n/2⌋
    (cantidad_estaciones_totales//2, cantidad_estaciones_totales)
]

#  mapas de calor para cada par
#  filas representan personas_iniciales (0 a capacidad) y las columnas personas_finales (0 a capacidad).
capacidad_maxima = capacidad  # de celda 6

for par in pares_estaciones:
    estacion_inicial, estacion_final = par
    # Crear la matriz de probabilidades donde:
    # fila = valor inicial personas_iniciales y columna = valor final personas_finales
    matriz_probabilidades = np.zeros((capacidad_maxima+1, capacidad_maxima+1))
    for personas_iniciales in range(capacidad_maxima+1):
        for personas_finales in range(capacidad_maxima+1):
            probabilidad = calcular_probabilidad(estacion_inicial, estacion_final, personas_iniciales, personas_finales)
            matriz_probabilidades[personas_iniciales, personas_finales] = probabilidad

    plt.figure(figsize=(9, 8))
    sns.heatmap(matriz_probabilidades, annot=True, fmt=".3f", cmap="viridis",
                xticklabels=range(capacidad_maxima+1), yticklabels=range(capacidad_maxima+1))
    plt.xlabel("Cantidad Final (personas_finales)")
    plt.ylabel("Cantidad Inicial (personas_iniciales)")
    plt.title(f"Mapa de calor de la probabilidad (Estación {estacion_inicial} → {estacion_final})")
    plt.show()
```

## Inciso c)

```python
def calcular_viajes_promedio(estacion_inicial, estacion_final, personas_iniciales, personas_finales):
    # Determinar la dirección inicial
    if estacion_final > estacion_inicial:
        direccion_inicial = 0
    else:
        direccion_inicial = 1
    estado_inicial = (personas_iniciales, estacion_inicial, direccion_inicial)
    
    memoizacion = {}
    def calcular_viajes_recursivo(estado_actual, visitados):
        if estado_actual in visitados:
            return float('inf')  # Evitar ciclos infinitos
        visitados.add(estado_actual)
        
        # Si llegamos al estado deseado
        if estado_actual[1] == estacion_final and estado_actual[0] == personas_finales:
            return 0
        
        if estado_actual in memoizacion:
            return memoizacion[estado_actual]
        
        viajes_acumulados = 0
        total_probabilidad = 0
        for siguiente_estado in estados:
            probabilidad_transicion = probabilidades_transicion.get((estado_actual, siguiente_estado), 0)
            if probabilidad_transicion > 0:
                viajes = calcular_viajes_recursivo(siguiente_estado, visitados.copy())
                if viajes != float('inf'):
                    viajes_acumulados += probabilidad_transicion * (viajes + 1)
                    total_probabilidad += probabilidad_transicion
        
        memoizacion[estado_actual] = viajes_acumulados / total_probabilidad if total_probabilidad > 0 else float('inf')
        return memoizacion[estado_actual]
    
    return calcular_viajes_recursivo(estado_inicial, set())

# Generar mapas de calor para los pares de estaciones
for par in pares_estaciones:
    estacion_inicial, estacion_final = par
    matriz_viajes_promedio = np.zeros((capacidad_maxima+1, capacidad_maxima+1))
    
    for personas_iniciales in range(capacidad_maxima+1):
        for personas_finales in range(capacidad_maxima+1):
            viajes_promedio = calcular_viajes_promedio(estacion_inicial, estacion_final, personas_iniciales, personas_finales)
            matriz_viajes_promedio[personas_iniciales, personas_finales] = viajes_promedio
    
    plt.figure(figsize=(9, 8))
    sns.heatmap(matriz_viajes_promedio, annot=False, fmt=".2f", cmap="coolwarm",
                xticklabels=range(capacidad_maxima+1), yticklabels=range(capacidad_maxima+1))
    plt.xlabel("Cantidad Final (personas_finales)")
    plt.ylabel("Cantidad Inicial (personas_iniciales)")
    plt.title(f"Mapa de calor de viajes promedio (Estación {estacion_inicial} → {estacion_final})")
    plt.show()
```

```python
import random
from collections import defaultdict

def simular_cmt(m):
    personas_iniciales = 0
    estacion_inicial = 1
    direccion_inicial = 0

    estado_actual = (personas_iniciales, estacion_inicial, direccion_inicial)
    historial_un_recorrido = []
    resultados_totales = defaultdict(list) 

    for simulacion in range(m):
        estado = estado_actual
        estaciones_recorridas = []
        viajes = 0

        while True:
            if simulacion == 0:
                historial_un_recorrido.append((viajes, estado[0], estado[1], estado[2]))
            estaciones_recorridas.append((estado[1], estado[0]))

            siguientes = [(siguiente_estado, prob) for ((origen, siguiente_estado), prob) in probabilidades_transicion.items() if origen == estado]
            if not siguientes:
                break

            estados_posibles, probabilidades = zip(*siguientes)
            estado = random.choices(estados_posibles, weights=probabilidades, k=1)[0]

            viajes += 1

            if estado[1] == 1 and estado[2] == 0:
                break

        for estacion, personas in estaciones_recorridas:
            resultados_totales[estacion].append(personas)

    return historial_un_recorrido, resultados_totales



historial, resultados = simular_cmt(m=10000)
# Gráfico de un recorrido
viajes = [h[0] for h in historial]
personas = [h[1] for h in historial]
estaciones = [h[2] for h in historial]
plt.figure(figsize=(12, 6))
plt.plot(viajes, personas, marker='o')
plt.xlabel('Viaje')
plt.ylabel('Cantidad de Personas')
plt.title('Evolución de personas en un recorrido')
plt.grid(True)
plt.show()
# Cantidad de gente promedio por estación
promedio_personas = {estacion: np.mean(valores) for estacion, valores in resultados.items()}
estaciones_ordenadas = sorted(promedio_personas.keys())
promedios = [promedio_personas[est] for est in estaciones_ordenadas]
plt.figure(figsize=(12, 6))
plt.bar(estaciones_ordenadas, promedios)
plt.xlabel('Estación')
plt.ylabel('Promedio de personas en el vagón')
plt.title('Promedio de personas al llegar a cada estación')
plt.grid(True)
plt.show()
# Distribución empírica (histograma) para las estaciones

for i in range(1, cantidad_estaciones+1):
    estacion_i = i
    M = 10
    plt.figure(figsize=(12, 6))
    plt.hist(resultados[estacion_i], bins=range(0, M+2), edgecolor='black', density=True)
    plt.xlabel('Cantidad de personas')
    plt.ylabel('Frecuencia relativa')
    plt.title(f'Distribución empírica de personas en estación {estacion_i}')
    plt.grid(True)
    plt.show()

```
