```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("DatosT4.xlsx")

```

1.Cantidad de personas que donaron a cada una de las fundaciones después del 1 de enero de 2021.

```python
df1 = df[df["Fecha de Donación"] >= pd.to_datetime("2021-01-01")].groupby(by = "Fundación").size()
print(df1)
```

Aldeas Infantiles SOS Chile: 44,
Coanil: 46,
DKMS Chile: 40,
Fundación Crescendo: 39,
Fundación Sonrisas: 32,
WWF Chile: 40

2.Gráfico de barras con la cantidad de donantes mayores a 30 años por Fundación.

```python
df2 = df[df["Edad"] >= 30].groupby(by = "Fundación").size()
df2.plot(kind="bar", color="purple")
print(df2)

plt.title("Cantidad de donantes mayores de 30 por Fundación")
plt.ylabel("Cantidad")
plt.xlabel("Fundación")


plt.show()
```

3.¿Cuál es la edad promedio de los donantes por tipo de frecuencia de participación?

```python
df3 = df.groupby("Frecuencia de Participación")["Edad"].mean()
print(df3)
```

Los donantes que participan anualmente tienen una edad promedio de 48,1 años; los que participan mensualmente, una edad promedio de 45,7 años; los que participan trimestralmente, 44,6 años; y, finalmente, quienes han participado una sola vez tienen una edad promedio de 46,4 años.

4.Porcentaje de donantes por sexo

```python
df4 = df.groupby(by = "Sexo").size()/len(df)
print(df4)
```

Un 45.8824% de los donantes son mujeres, mientras que un 54.1176% son hombres.

5.¿Cuál es la razón de donación más frecuente en DKMS Chile?

```python
df5 = df[df["Fundación"] == "DKMS Chile"].groupby(by="Razón Tras la Donación").size()
print(df5)
print(f"Razón más frecuente: {df5.idxmax()} ({df5.max()} donantes)")
```

La razón de donación mas frecuente en DKMS Chile es la experiencia personal, pues 10 personas han donado en DKMS por ese motivo.

6.¿Cuántas personas donaron desde México y a qué fundaciones?

```python

print(f"Total de personas que donaron desde México: {len(df[df['País de Origen'] == 'México'])}")
df6 = df[df["País de Origen"] == "México"].groupby(by="Fundación").size()
print(df6)
```

7.¿Qué fundación tiene la mayor proporción de donantes que contribuyen de forma “mensual”? Además, mostrar las proporciones de cada fundación. En esta pregunta no se especifica si es la proporción de donantes sobre el total (independiente de la frecuencia) o  sobre el total de personas que donan mensualmente.

Proporción sobre el total de personas que donan mensualmente:

```python
df7 = df[df["Frecuencia de Participación"] == "mensual"].groupby(by="Fundación").size()/len(df[df["Frecuencia de Participación"] == "mensual"])
print(df7)
```

Como se puede ver, la fundación con la mayor proporción de donantes (sobre el total de personas que donan mensualmente) que contribuyen de forma mensual es DKMS Chile, con un 21.9512% de un total de 82 donantes.
